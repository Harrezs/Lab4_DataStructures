```python
#EXERCISE NUMBER 1
import random 
from collections import defaultdict

LAST_NAME = "Belga"
STUDENT_ID = "TUPM-26-1240"
FAVORITE_ARTIST = "Chase Atlantic"
SEED_NUM = int(STUDENT_ID[-1]) + len(LAST_NAME)

random.seed(SEED_NUM)

test_sources = [f"{LAST_NAME}_T1", f"{LAST_NAME}_T2", f"{LAST_NAME}_T3", f"{LAST_NAME}_T4", f"{LAST_NAME}_T5"]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

test_records = []
for i in range(10):
    source = test_sources[i % len(test_sources)]
    cat = categories[(SEED_NUM + i) % len(categories)]
    char_val = ord(FAVORITE_ARTIST[i % len(FAVORITE_ARTIST)])

    val = round(((SEED_NUM * 5) + char_val + (i * 3)) % 150 + 10.5, 2)
    test_records.append({
        "source": source,
        "category": cat,
        "value": val
    })

test_records.append(test_records[1].copy())

print("=== 1. GENERATED TEST DATA ===")
for rec in test_records:
    print(rec)

seen_records = set()
repeated_records = []
distinct_records = []

for rec in test_records:
    rec_tuple = (rec["source"], rec["category"], rec["value"])
    if rec_tuple in seen_records:
        repeated_records.append(rec)
    else:
        seen_records.add(rec_tuple)
        distinct_records.append(rec)

print(f"\n=== 3. REPEATED & DISTINCT DATA SUMMARY ===")
print(f"Total Raw Records Generated: {len(test_records)}")
print(f"Repeated Records Found: {len(repeated_records)}")
print(f"Distinct Records Count: {len(distinct_records)}")

organized_by_source = defaultdict(list)
organized_by_category = defaultdict(list)

for rec in distinct_records:
    organized_by_source[rec["source"]].append(rec["value"])
    organized_by_category[rec["category"]].append(rec["value"])

all_values = [rec["value"] for rec in distinct_records]
overall_avg = sum(all_values) / len(all_values)
max_val = max(all_values)
min_val = min(all_values)

ordered_results = sorted(distinct_records, key=lambda x: x["value"], reverse=True)

print("\n" + "="*50)
print("     FINAL ORGANIZED AND SUMMARIZED TEST DATASET     ")
print("="*50)
print(f"Engineer/Operator: {LAST_NAME} | Student ID: {STUDENT_ID}")
print(f"Overall Dataset Min: {min_val} | Max: {max_val} | Avg: {overall_avg:.2f}")

print("\n--- Average Measurement Values by Category ---")
for cat, vals in organized_by_category.items():
    avg = sum(vals) / len(vals)
    print(f"{cat}: {avg:.2f} (Readings Count: {len(vals)})")

print("\n--- Average Measurement Values by Test Source ---")
for src, vals in organized_by_source.items():
    avg = sum(vals) / len(vals)
    print(f"{src}: {avg:.2f}")

print("\n--- Processed Results (Highest Recorded Values) ---")
for rec in ordered_results[:5]:
    print(f"[{rec['source']}] {rec['category']}: {rec['value']}")
```

```python
#EXERCISE NUMBER 2
import random
LAST_NAME = "Belga"
STUDENT_ID = "TUPM-26-1240"
FAVORITE_ARTIST = "Chase Atlantic"
SEED_NUM = int(STUDENT_ID[-1]) + len(LAST_NAME)

random.seed(SEED_NUM)

students = [f"{LAST_NAME}_S1", f"{LAST_NAME}_S2", f"{LAST_NAME}_S3", f"{LAST_NAME}_S4", f"{LAST_NAME}_S5"]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH" ]

performance_records = []
idx = 0
for student in students:
    for subject in subjects[:3]:
        char_val = ord(FAVORITE_ARTIST[idx % len(FAVORITE_ARTIST)])
        score = min(100, max(50, (SEED_NUM * 7 + char_val + idx * 3) % 51 + 50))
        performance_records.append({
            "student": student,
            "subject": subject,
            "score": score
        })
        idx += 1

performance_records.append(performance_records[0].copy())

performance_records.append({
    "student": students[1],
    "subject": "ENGLISH",
    "score": None
})

print("=== 1. GENERATED PERFORMANCE RECORDS ===")
for rec in performance_records:
    print(rec)

seen_records = set()
repeated_records = []
incomplete_records = []

for rec in performance_records:
    if rec["score"] is None:
        incomplete_records.append(rec)
        continue

    rec_tuple = (rec["student"], rec["subject"], rec["score"])
    if rec_tuple in seen_records:
        repeated_records.append(rec)
    else:
        seen_records.add(rec_tuple)

def assign_category(score):
    if score >= 90:
        return "EXCELLENT"
    elif score >= 75:
        return "SATISFACTORY"
    else:
        return "NEEDS IMPROVEMENT"

valid_records = []
seen = set()
for rec in performance_records:
    if rec["score"] is not None:
        rec_tuple = (rec["student"], rec["subject"], rec["score"])
        if rec_tuple not in seen:
            seen.add(rec_tuple)
            rec["category"] = assign_category(rec["score"])
            valid_records.append(rec)

student_summary = {}
subject_summary = {}

for rec in valid_records:
    stu = rec["student"]
    subj = rec["subject"]
    scr = rec["score"]

    if stu not in student_summary:
        student_summary[stu] = []
    student_summary[stu].append(scr)

    if subj not in subject_summary:
        subject_summary[subj] = []
    subject_summary[subj].append(scr)

ordered_records = sorted(valid_records ,key=lambda x: x["score"],  reverse=True)

print("\n" + "="*50)
print("     FINAL STUDENT PERFORMANCE REPORT     ")
print("="*50)
print(f"Instructor/Operator: {LAST_NAME} | Student ID: {STUDENT_ID}")
print(f"Total EvaluatedRecords: {len(performance_records)}")
print(f"Repeated Records Identified: {len(repeated_records)}")
print(f"Incomplete Records Identified: {len(incomplete_records)}")

print("\n--- Student Average Scores ---")
for stu, scores in student_summary.items():
    avg = sum(scores) / len(scores)
    print(f"{stu}: {avg:.2f}")

print("\n--- Subject Average Scores ---")
for subj, scores in subject_summary.items():
    avg = sum(scores) / len(scores)
    print(f"{subj}: {avg:.2f}")

print("\n--- Top Ranked Student Performances ---")
for rec in ordered_records[:5]:
    print(f"[{rec['student']}] {rec['subject']}: {rec['score']} ({rec['category']})")
```

```python

# EXERCISE 3 - TRANSACTION HISTORY ANALYZER
LAST_NAME = "Belga"
STUDENT_ID = "TUPM-26-1240"
SEED_NUM = 1240
FAVORITE_ARTIST = "Chase Atlantic"

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

products = {
    "NOTEBOOK": 35,
    "PEN": 15,
    "USB": 250,
    "HEADPHONES": 450,
    "CALCULATOR": 180
}

purchase_list = [
    ("NOTEBOOK", 0), ("PEN", 0), ("USB", 0),
    ("PEN", 1), ("HEADPHONES", 1), ("CALCULATOR", 1),
    ("USB", 2), ("CALCULATOR", 2), ("NOTEBOOK", 2),
    ("HEADPHONES", 3), ("NOTEBOOK", 3), ("PEN", 3),
    ("CALCULATOR", 4), ("USB", 4), ("HEADPHONES", 4)
]

artist_values = [
    ord(letter.lower()) - 96
    for letter in FAVORITE_ARTIST if letter.isalpha()
]

transactions = []

for i, (product, customer_index) in enumerate(purchase_list):
    quantity = (SEED_NUM + artist_values[i % len(artist_values)] + i) % 5 + 1

    transaction = {
        "transaction_id": f"T{i + 1:02d}",
        "customer": customers[customer_index],
        "product": product,
        "quantity": quantity,
        "unit_price": products[product],
        "total": quantity * products[product]
    }

    transactions.append(transaction)

repeated_purchase = transactions[0].copy()
repeated_purchase["transaction_id"] = "T16"
transactions.append(repeated_purchase)

print("=== GENERATED TRANSACTION RECORDS ===")

for transaction in transactions:
    print(transaction)

customer_data = {}
product_data = {}

for customer in customers:
    customer_data[customer] = [
        t for t in transactions if t["customer"] == customer
    ]

for product in products:
    product_data[product] = [t for t in transactions if t["product"] == product]

pair_counts = {}

for t in transactions:
    pair = (t["customer"], t["product"])
    pair_counts[pair] = pair_counts.get(pair, 0) + 1

repeated_purchases = {
    pair: count for pair, count in pair_counts.items() if count > 1
}

customer_summary = {}

for customer, records in customer_data.items():
    customer_summary[customer] = {
        "transactions": len(records),
        "total_quantity": sum(t["quantity"] for t in records),
        "total_spent": sum(t["total"] for t in records)
    }

product_summary = {}

for product, records in product_data.items():
    product_summary[product] = {
        "transactions": len(records),
        "total_quantity": sum(t["quantity"] for t in records),
        "total_sales": sum(t["total"] for t in records)
    }

ordered_transactions = sorted(
    transactions,
    key=lambda t: t["total"],
    reverse=True
)

total_sales = sum(t["total"] for t in transactions)
total_quantity = sum(t["quantity"] for t in transactions)

print("\n=== REPEATED CUSTOMER-PRODUCT PURCHASES ===")
for pair, count in repeated_purchases.items():
    print(pair, "- Occurrences:", count)

print("\n=== CUSTOMER SUMMARIES ===")
for customer, summary in customer_summary.items():
    print(customer, summary)

print("\n=== PRODUCT SUMMARIES ===")
for product, summary in product_summary.items():
    print(product, summary)

print("\n=== ORDERED TRANSACTION RESULTS ===")
for transaction in ordered_transactions:
    print(transaction)

print("\n=== OVERALL TRANSACTION SUMMARY ===")
print("Total transactions:", len(transactions))
print("Distinct customer-product pairs:", len(pair_counts))
print("Number of repeated pairs:", len(repeated_purchases))
print("Total quantity purchased:", total_quantity)
print("Total transaction value: PHP", total_sales)

print("\n=== FINAL CONSOLIDATED TRANSACTION REPORT ===")
print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Total transaction records:", len(transactions))
print("Number of customers:", len(customers))
print("Number of products:", len(products))
print("Total sales: PHP", total_sales)

top_customer = max(
    customer_summary,
    key=lambda c: customer_summary[c]["total_spent"]
)

top_product = max(
    product_summary,
    key=lambda p: product_summary[p]["total_sales"]
)

print("Highest spending customer:", top_customer)
print("Best-selling product by sales value:", top_product)
print("Repeated purchases detected:", len(repeated_purchases))
```
