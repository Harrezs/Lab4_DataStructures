```python
#EXERCISE NUMBER 1
import random 
from collections import defaultdict

LAST_NAME = "Belga"
STUDENT_ID = "TUPM-26-1240"
FAVORITE_ARTIST = "Chase Atlantic"
SEED_NUM = int(STUDENT_ID[-1]) + len(LAST_NAME)

random.seed(SEED_NUM)

test_sources = [f"{LAST_NAME}_T1", f"{LAST_NAME}_T2", f"{LAST_NAME}_T3", f"{LAST_NAME}_T4", f"{LAST_NAME}_T5"]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

test_records = []
for i in range(10):
    source = test_sources[i % len(test_sources)]
    cat = categories[(SEED_NUM + i) % len(categories)]
    char_val = ord(FAVORITE_ARTIST[i % len(FAVORITE_ARTIST)])

    val = round(((SEED_NUM * 5) + char_val + (i * 3)) % 150 + 10.5, 2)
    test_records.append({
        "source": source,
        "category": cat,
        "value": val
    })

test_records.append(test_records[1].copy())

print("=== 1. GENERATED TEST DATA ===")
for rec in test_records:
    print(rec)

seen_records = set()
repeated_records = []
distinct_records = []

for rec in test_records:
    rec_tuple = (rec["source"], rec["category"], rec["value"])
    if rec_tuple in seen_records:
        repeated_records.append(rec)
    else:
        seen_records.add(rec_tuple)
        distinct_records.append(rec)

print(f"\n=== 3. REPEATED & DISTINCT DATA SUMMARY ===")
print(f"Total Raw Records Generated: {len(test_records)}")
print(f"Repeated Records Found: {len(repeated_records)}")
print(f"Distinct Records Count: {len(distinct_records)}")

organized_by_source = defaultdict(list)
organized_by_category = defaultdict(list)

for rec in distinct_records:
    organized_by_source[rec["source"]].append(rec["value"])
    organized_by_category[rec["category"]].append(rec["value"])

all_values = [rec["value"] for rec in distinct_records]
overall_avg = sum(all_values) / len(all_values)
max_val = max(all_values)
min_val = min(all_values)

ordered_results = sorted(distinct_records, key=lambda x: x["value"], reverse=True)

print("\n" + "="*50)
print("     FINAL ORGANIZED AND SUMMARIZED TEST DATASET     ")
print("="*50)
print(f"Engineer/Operator: {LAST_NAME} | Student ID: {STUDENT_ID}")
print(f"Overall Dataset Min: {min_val} | Max: {max_val} | Avg: {overall_avg:.2f}")

print("\n--- Average Measurement Values by Category ---")
for cat, vals in organized_by_category.items():
    avg = sum(vals) / len(vals)
    print(f"{cat}: {avg:.2f} (Readings Count: {len(vals)})")

print("\n--- Average Measurement Values by Test Source ---")
for src, vals in organized_by_source.items():
    avg = sum(vals) / len(vals)
    print(f"{src}: {avg:.2f}")

print("\n--- Processed Results (Highest Recorded Values) ---")
for rec in ordered_results[:5]:
    print(f"[{rec['source']}] {rec['category']}: {rec['value']}")
```

```python
#EXERCISE NUMBER 2
import random
LAST_NAME = "Belga"
STUDENT_ID = "TUPM-26-1240"
FAVORITE_ARTIST = "Chase Atlantic"
SEED_NUM = int(STUDENT_ID[-1]) + len(LAST_NAME)

random.seed(SEED_NUM)

students = [f"{LAST_NAME}_S1", f"{LAST_NAME}_S2", f"{LAST_NAME}_S3", f"{LAST_NAME}_S4", f"{LAST_NAME}_S5"]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH" ]

performance_records = []
idx = 0
for student in students:
    for subject in subjects[:3]:
        char_val = ord(FAVORITE_ARTIST[idx % len(FAVORITE_ARTIST)])
        score = min(100, max(50, (SEED_NUM * 7 + char_val + idx * 3) % 51 + 50))
        performance_records.append({
            "student": student,
            "subject": subject,
            "score": score
        })
        idx += 1

performance_records.append(performance_records[0].copy())

performance_records.append({
    "student": students[1],
    "subject": "ENGLISH",
    "score": None
})

print("=== 1. GENERATED PERFORMANCE RECORDS ===")
for rec in performance_records:
    print(rec)

seen_records = set()
repeated_records = []
incomplete_records = []

for rec in performance_records:
    if rec["score"] is None:
        incomplete_records.append(rec)
        continue

    rec_tuple = (rec["student"], rec["subject"], rec["score"])
    if rec_tuple in seen_records:
        repeated_records.append(rec)
    else:
        seen_records.add(rec_tuple)

def assign_category(score):
    if score >= 90:
        return "EXCELLENT"
    elif score >= 75:
        return "SATISFACTORY"
    else:
        return "NEEDS IMPROVEMENT"

valid_records = []
seen = set()
for rec in performance_records:
    if rec["score"] is not None:
        rec_tuple = (rec["student"], rec["subject"], rec["score"])
        if rec_tuple not in seen:
            seen.add(rec_tuple)
            rec["category"] = assign_category(rec["score"])
            valid_records.append(rec)

student_summary = {}
subject_summary = {}

for rec in valid_records:
    stu = rec["student"]
    subj = rec["subject"]
    scr = rec["score"]

    if stu not in student_summary:
        student_summary[stu] = []
    student_summary[stu].append(scr)

    if subj not in subject_summary:
        subject_summary[subj] = []
    subject_summary[subj].append(scr)

ordered_records = sorted(valid_records ,key=lambda x: x["score"],  reverse=True)

print("\n" + "="*50)
print("     FINAL STUDENT PERFORMANCE REPORT     ")
print("="*50)
print(f"Instructor/Operator: {LAST_NAME} | Student ID: {STUDENT_ID}")
print(f"Total EvaluatedRecords: {len(performance_records)}")
print(f"Repeated Records Identified: {len(repeated_records)}")
print(f"Incomplete Records Identified: {len(incomplete_records)}")

print("\n--- Student Average Scores ---")
for stu, scores in student_summary.items():
    avg = sum(scores) / len(scores)
    print(f"{stu}: {avg:.2f}")

print("\n--- Subject Average Scores ---")
for subj, scores in subject_summary.items():
    avg = sum(scores) / len(scores)
    print(f"{subj}: {avg:.2f}")

print("\n--- Top Ranked Student Performances ---")
for rec in ordered_records[:5]:
    print(f"[{rec['student']}] {rec['subject']}: {rec['score']} ({rec['category']})")
```
